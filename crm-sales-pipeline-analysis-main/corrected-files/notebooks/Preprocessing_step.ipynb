```python
import pandas as pd
```

```python
# Load files
accounts = pd.read_csv('accounts.csv')
products = pd.read_csv('products.csv')
sales_pipeline = pd.read_csv('sales_pipeline.csv')
sales_teams = pd.read_csv('sales_teams.csv')
```

```python

# Handle nulls (sensible defaults)
accounts['subsidiary_of'] = accounts['subsidiary_of'].fillna('Independent')  # Or 'None'
sales_pipeline['account'] = sales_pipeline['account'].fillna('Unassigned')
sales_pipeline['engage_date'] = pd.to_datetime(sales_pipeline['engage_date'], errors='coerce')  # Fix dates
sales_pipeline['close_date'] = pd.to_datetime(sales_pipeline['close_date'], errors='coerce')
sales_pipeline['close_value'] = sales_pipeline['close_value'].fillna(0)  # For open/lost deals
```

```python

# Add useful calculated columns (for KPIs)
sales_pipeline['sales_cycle_days'] = (sales_pipeline['close_date'] - sales_pipeline['engage_date']).dt.days.fillna(0)
sales_pipeline['is_won'] = (sales_pipeline['deal_stage'] == 'Won').astype(int)
sales_pipeline['year'] = sales_pipeline['close_date'].dt.year.fillna(0)  # Or engage_date
```

```python

# Save cleaned versions
accounts.to_csv('accounts_clean.csv', index=False)
products.to_csv('products_clean.csv', index=False)
sales_pipeline.to_csv('sales_pipeline_clean.csv', index=False)
sales_teams.to_csv('sales_teams_clean.csv', index=False)
```
